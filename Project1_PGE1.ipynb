<h3>Overview</h3>
<p>The application consists of several functionalities divided into user and admin roles.</p>
<ol>
    <li><h5>User role: </h5>
    <ul>
        <li> Users can log in, view the product catalog, add/remove items from their shopping cart, and proceed to checkout.</li>
    </ul>
     </li>
    <li>
        <h5> Admin role: </h5>
        <ul>
            <li> Admins can log in, view the product catalog, add new products, or remove existing products. </li>
        </ul>
    </li>
</ol>

<h3>Data Structures</h3>
<ol>
    <li><b>user_db</b>: A dictionary storing user credentials (username-password pairs).</li>
    <li><b>admin_db</b>: A dictionary storing admin credentials (username-password pairs).</li>
    <li><b>product_catalog</b>: A list of dictionaries representing the available products in the marketplace. Each dictionary contains:
    <ul>
        <li>Product_id: Unique ID for the product </li>
        <li>Name: Product name </li>
        <li>Category_id: Product Category_id(according to clothing, shoes, pants) </li>
        <li>price: Price of the product </li>
    </ul>
    </li>
    <li> <b>sessions</b>: A dictionary that tracks active sessions for logged-in users and admins</li>
    <li> <b>carts</b>: A dictionary where each session ID maps to a shopping cart containing selected products and their quantities</li>
</ol>

```python
# Backend Shopping App

# Demo Databases

user_db= {'user1': 'pass1','user2': 'pass2'}
admin_db= {'admin': 'admin1'}

product_catalog = [
    {"product_id": "P001", "name": "Boots", "category_id": "1", "price": 2500},
    {"product_id": "P002", "name": "Coats", "category_id": "2", "price": 3500},
    {"product_id": "P003", "name": "Jackets", "category_id": "2", "price": 4500},
    {"product_id": "P004", "name": "Caps", "category_id": "3", "price": 500}
]
# Active sessions and carts
sessions = {}
carts = {}
session_counter = 1
```

<h3>Key Functions</h3>
<ul>
    <li><b>welcome_message()</b></li>
    <p>Displays a welcome message to users upon launching the application</p>
</ul>

```python
# Display welcoming message

def welcome_message():
    welcome_msg= '''
                             -----------------------------------
                             | Welcome to the Demo Marketplace |
                             -----------------------------------
    '''
    print(welcome_msg)
```

<ul>
    <li>
        <b> view_catalog() </b>
    </li>
    <p>Displays the list of products available in the marketplace in their details( Id,name,category, and price).</p>
</ul>

```python
# View Catalog : view all products with their description

def view_catalog():
    print("\nProduct Catalog:")
    for product in product_catalog:
        print(f"ID: {product['product_id']}, Name: {product['name']}, Category_ID: {product['category_id']}, Price: Rs. {product['price']}")
    
    
```

<ul>
    <li>
        <b> manage_cart(session_id) </b>
    </li>
    <br>
    <p>Allows users to:</p>
    <ul>
        <li>View items in their cart.</li>
        <li>Add or remove products from their cart.</li>
        <li>Proceed to checkout using payment options(UPI, Debit Card, or Net Banking).</li>
    </ul>
    <br>
    <p>Key steps in cart management</p>
    <ul>
        <li>Products are identified using <b>product_id</b></li>
        <li>Total cose is calculated by iterating through the cart and multiplying product prices by quantities.</li>
    </ul>
</ul>

```python
# manage cart
def manage_cart(session_id):
    user_cart= carts.setdefault(session_id,{})
    choose= '''
    ---------------
    | Cart Options |
    ---------------

    
    1.| View Cart
    --------------------------
    2.| Add Item to Cart
    --------------------------
    3.| Remove Item from Cart
    --------------------------
    4.| Checkout
    --------------------------
    5.| Logout
    --------------------------
    '''
    while True:
        print(choose)
        choice= input('Enter your choice: ')
        if choice=="1":
            print('\n Your Cart:')
            if not user_cart:
                print('Your cart is empty')
            else:
                for product_id,quantity in user_cart.items():
                    for product in product_catalog:
                        if product['product_id']==product_id:
                            print(f"{product['name']} - Quantity: {quantity} - Price: Rs. {product['price'] * quantity}")
                            break
        elif choice=='2':
            view_catalog()
            product_id= input('Enter the product ID to add: ')
            quantity= int(input("Enter the quantity: "))
            if any(p['product_id']==product_id for p in product_catalog):
                user_cart[product_id] = user_cart.get(product_id,0)+quantity
                print('\nItem added to cart.')
            else:
                print('Invalid Product ID.')

        elif choice=='3':
            product_id= input('Enter the product ID to remove: ')
            if product_id in user_cart:
                del user_cart[product_id]
                print('Item removed from cart.')
            else:
                print('Product not found in your cart.')
        elif choice=='4':
            options= '''
               --------------------
               | Checkout options |
               --------------------

 
            1.| UPI
            ---------------
            2.| Debit Card
            ---------------
            3.| Net Banking
            '''
            print(options)
            payment_choice= input('Choose a payment option: ')
            if payment_choice in ['1','2','3']:
                total=0
                for product_id, quantity in user_cart.items():
                    for product in product_catalog:
                        if product['product_id']==product_id:
                            total+=product['price'] *quantity
                            break
                print(f'Your order is successfully placed. Total amount: Rs {total}')
                carts[session_id]={}
                return
            else:
                print('Invalid payment option.')
            
        elif choice=='5':
            print('Logging out..')
            return
        else:
            print('Invalid Choice. Try again.')
                              
            
    

```

<ul>
    <li><b>login_user(role)</b></li>
    <p>Handles login for users and admins: </p>
    <ul>
        <li>Validates credentials.</li>
        <li>Generates a unique session ID upon successful login.</li>
        <li>Returns the session ID for further actions.</li>
    </ul>
</ul>

```python
# User Login function
def login(role):
    global session_counter
    username = input(f"Enter {role} username: ")
    password= input('Enter password')
    if role=='user' and user_db.get(username)==password:
        session_id= f'session_{session_counter}'
        session_counter+=1
        sessions[session_id]=username
        print(f'{username} is logged in. Your Session Id is {session_id}')
        return session_id
    elif role=='admin' and admin_db.get(username)==password:
        session_id= f'session_{session_counter}'
        session_counter+=1
        sessions[session_id]=username
        print(f'{username} is logged in. Your Session Id is {session_id}')
        return session_id
    else:
        print("Invalid credentials.")
        return None
        
    
        
```

<ul>
    <li>
        <b>admin_actions(session_id)</b>
    </li>
    <br>
    <p>Enables admins to: </p>
    <ul>
        <li>View the product catalog.</li>
        <li>Add new products to the catalog by providing name, category, and price.</li>
        <li>Remove the products using their <b>product_id</b></li>
    </ul>
</ul>

```python
def admin_actions(session_id):
    
    choose= '''
    ---------------
    | Admin Options |
    ---------------

    
    1.| View Product Catalog
    --------------------------
    2.| Add Product
    --------------------------
    3.| Remove Product
    --------------------------
    4.| Logout
    --------------------------
    '''
    while True:
        print(choose)
        choice = input('Enter your choice: ')
        if choice == '1':
            view_catalog()
        elif choice == '2':
            name = input("Enter product name: ")
            category = input("Enter product category: ")
            price = int(input("Enter product price: "))
            product_id = f"P{len(product_catalog) + 1:03}"
            product_catalog.append({"product_id": product_id, "name": name, "category_id": category, "price": price})
            print("Product added successfully.")
        elif choice == '3':
            product_id= input("Enter the Product ID to remove: ")
            product_catalog[:] = [p for p in product_catalog if p['product_id']!= product_id]
            print('Product removed successfully')
        elif choice =='4':
            print('Logging out..')
            return
        else:
            print("Invalid option. Try again.")
                  
    
```

<ul>
    <li>
        <b>main() </b>
    </li>
    <p>The main entry point of the application:</p>
    <ul>
        <li>Displays the main menu for users to select between User Login, Admin Login, or Exit.</li>
        <li>Direct logged-in users to either cart management (<b>manage_cart</b>) or admin actions (<b>admin_actions</b>) based on their role.</li>
    </ul>
</ul>

```python
def main():
    welcome_message()
    options= '''
    -------------
    | Main Menu |
    -------------

 
    1.| User Login
    ---------------
    2.| Admin Login
    ---------------
    3.| Exit
   
    
    '''
    while True:
        print(options)
        choice= input('Enter your choice:')
        if choice=='1':
            session_id=login("user")
            if session_id:
                manage_cart(session_id)
        elif choice=='2':
            session_id= login('admin')
            if session_id:
                admin_actions(session_id)
        elif choice=='3':
            print('\nThank You for visiting the Demo Marketplace\n')
            break
        else:
            print("\n****Invalid choice. Please Try again.****\n")
```

```python
# ****IMPORTANT****
# Program starts here
if __name__== "__main__":
   main()
```
